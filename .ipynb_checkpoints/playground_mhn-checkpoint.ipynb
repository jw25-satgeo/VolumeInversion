```python
import sys, os, datetime, importlib

import pandas as pd
import numpy as np
import scipy
import matplotlib.pyplot as plt

import torch
import torch.nn.functional as F

import torchvision
read = torchvision.datasets.CIFAR10(root='ds_CIFAR10', train=True, download=False)
read = torch.as_tensor(read.data[:200]).clone()
```

```python
import nn_ebm; importlib.reload(ebm)
loss_hist = []
train = read[:20,8:16,8:16,0].to(torch.float32)/255
model = nn_ebm.dynEBAM(read.flatten(1).shape[-1], 128, lr=0.007, r_sigma=1e-4, dist=nn_ebm.dist_dp).to(torch.float32)
model.learn(train.flatten(1), beta0=1, iters=10, c=7, loss_hist=loss_hist)

plt.plot(loss_hist); plt.show()
```

```python
feed = train + 0.5*torch.rand_like(train)
result = model(feed.flatten(1), beta0=1, iters=5).detach().cpu().view(train.shape)
#plt.imshow(result[0]); plt.show()
for i in range(result.shape[0]):
    plt.subplot(1,3,1); plt.imshow(feed[i]);
    plt.subplot(1,3,2); plt.imshow(result[i]);
    plt.subplot(1,3,3); plt.imshow(train[i])
    plt.show()
    
    
```
