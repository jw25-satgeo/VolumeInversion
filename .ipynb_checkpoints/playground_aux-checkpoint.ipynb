```python
import crefl2slc; importlib.reload(crefl2slc)
srefi = crefl2slc.SimpleReflector.load()

c_slc = torch.as_tensor(slc_ds.slc[:,:500], device=srefi.device, dtype=torch.complex64)
c_slc /= torch.max(torch.abs(c_slc)) # normalize magnitude
#c_slc = torch.stack((c_slc[0] + c_slc[3], c_slc[0] - c_slc[3], c_slc[1] + c_slc[2]), dim=0)
c_ia = torch.as_tensor(slc_ds.ia[:500], device=srefi.device, dtype=srefi.dtype)
c_rg = torch.as_tensor(slc_ds.rg[:500], device=srefi.device, dtype=srefi.dtype)

p_thc = torch.ones_like(c_slc[0], dtype=srefi.dtype, requires_grad=True)
p_T = torch.tensor([[-2.5]], device=srefi.device, dtype=srefi.dtype, requires_grad=True)
p_scale = torch.tensor([10], device=srefi.device, dtype=srefi.dtype, requires_grad=True) # rescale offset

optimizer = torch.optim.AdamW([p_thc, p_T, p_scale], lr=0.1)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=50)
for i in range(50):
    p_slc = srefi.slc_simulate(p_thc, c_rg, c_ia, p_T.expand(p_thc.shape), F.relu(p_scale)).permute(2,0,1)
    #p_slc = torch.stack(p_slc[0] - p_slc[3], p_slc[1] + p_slc[2]), dim=0)
    loss = (torch.abs(p_slc - c_slc)**2).mean() + ((torch.abs(p_slc) - torch.abs(c_slc))**2).mean()
    loss += 0.2 * ((p_thc[1:,:-1] - p_thc[:-1,:-1])**2 + (p_thc[:-1,1:] - p_thc[:-1,:-1])**2).mean() # TV loss
    loss += torch.exp(-100*p_thc).sum() # clamp to positive value
    print(i, loss.item(), 'T_est:', p_T.item(), p_scale.item())
    loss.backward()
    optimizer.step()
    scheduler.step()
    optimizer.zero_grad()
#del(optimizer, c_slc, c_ia, c_rg, p_slc)
```
