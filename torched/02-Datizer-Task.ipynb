```python
import os
import PyPreprocess
import PolarDecomp
import matplotlib.pyplot as plt
import torch
import numpy as np
```

```python
root_dir = '/mnt/c/Users/user/Desktop/slc_cache/251229r' #Users\user\Desktop\slc_cache\251229r
pair_dirs = [os.path.join(root_dir, el) for el in os.listdir(root_dir) if '.' not in el]
```

```python
coh_labels = ['HH','HV','VH','VV']
counter = 0
for pair_dir in pair_dirs:
    #if counter < 9: counter+=1; continue
    fn_1, fn_2 = [os.path.join(pair_dir, el) for el in os.listdir(pair_dir) if '.zip' in el]
    #dataset = PyPreprocess.Pair_Dataset_SNAPNative(s1_fn=fn_1, s2_fn=fn_2, cache_dir=pair_dir, read_nc=False)
    dataset = PyPreprocess.Pair_Dataset_SNAPNative(s1_fn=fn_1, s2_fn=fn_2, cache_dir=pair_dir, read_nc=True)
    print(dataset.baseline)
    break
    qpol1, qpol2 = dataset.s1, dataset.s2
    baseline = dataset.baseline
    del(dataset)
    if (np.abs(qpol1).min() == 0.0):
        with open(f'{pair_dir}/error.txt', 'w', encoding='utf-8') as f:
            f.write(f'bad coregistration - SNAPISTA produced blank s2 image. The perpendicular baseline was {baseline},
            which may indicate the validity of the pair.')
        continue
    coherences = PolarDecomp.CoherenceBuilder(qpol1, qpol2, dtype=torch.complex64, tensor=False)[0]
    for i in range(coherences.shape[-1]):
        plt.imshow(np.abs(coherences[:,:,i].T), aspect=5, vmin=0.0, vmax=1.0);
        plt.savefig(f'{pair_dir}/{coh_labels[i]}_coherence_magnitude.png')
        plt.imshow(np.angle(coherences[:,:,i].T), aspect=5, vmin=0.0, vmax=1.0);
        plt.savefig(f'{pair_dir}/{coh_labels[i]}_coherence_phase.png')
    counter += 1
```
