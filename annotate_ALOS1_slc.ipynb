```python
import sys, os, importlib, tqdm
import pickle

import torch
import torch.nn.functional as F
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import netCDF4, h5py, scipy
```

```python
import PyPreprocess_Single
from PIL import Image

slc_dir = '/mnt/d/ALOS1-Landfast/BeaufortSouth'
slc_fn_l = [os.path.join(slc_dir, el) for el in os.listdir(slc_dir) if el[-4:] == '.zip']

for slc_fn in slc_fn_l:
    cache_dir = slc_fn[:-4] + '_cache'
    if not os.path.exists(cache_dir):
        try:
            slc_ds = PyPreprocess_Single.Dataset_SNAPbase(src_fn=slc_fn, cache_dir=cache_dir, read_nc=False, alos1_deskew=True)
            paulix = np.abs(np.stack([ # (H,W,3) Pauli basis magnitude
                slc_ds.slc[0] + slc_ds.slc[3], # hh + vv
                slc_ds.slc[0] - slc_ds.slc[3], # hh - vv
                slc_ds.slc[1] + slc_ds.slc[2], # hv + vh
            ], axis=-1)) / np.sqrt(2)
            # rescale values to 0~255 for image write
            paulix = (255 * (paulix / paulix.max(axis=(0, 1), keepdims=True))).astype(np.uint8)
            # export pauli basis preview & empty mask for editing
            Image.fromarray(paulix, mode='RGB').save(os.path.join(cache_dir, 'preview_pauli.png'))
            Image.fromarray(np.zeros(paulix.shape[:2], dtype=np.uint8), mode='L').save(os.path.join(cache_dir, 'mask_base.png'))
        except:
            continue
    #else:
    #    mask = np.array(Image.open(os.path.join(cache_dir, 'mask.png')), dtype=np.uint8) > 127
    
```
