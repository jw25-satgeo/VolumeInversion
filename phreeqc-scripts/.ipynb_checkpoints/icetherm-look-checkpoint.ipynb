```python
import phreeqpy.iphreeqc.phreeqc_dll as phreeqc_mod

#dir(phreeqc_mod.IPhreeqc)
phreeqc = phreeqc_mod.IPhreeqc('/usr/local/lib/libiphreeqc.so')
#os.listdir("frezchem-master")
```

```python
#phreeqc.load_database("frezchem-master/frezchemSiCH4.dat")
phreeqc.load_database("/home/adminix/iphreeqc-3.8.6-17100/database/frezchem.dat")
phreeqc.phc_database_error_count 
```

```python
ref_molmasses = np.array([22.98977, 35.453, 24.305, 96.0626, 40.078, 39.0983]) #g/mol
ref_solutes = np.array([0.468, 0.545, 0.052, 0.028, 0.010, 0.010]) #(mol/kgw) @ ~35.2 g/kg salinity - adds up to 34.8265 g/kg

# - https://faculty.uml.edu/david_ryan/84.653/Course%20Material/lecture10/Dissolved%20gases%202011.pdf
#ref_seagas = np.array([20, 349, 616])*1e-6 #CO2, O2, N2 fractions - umol/kg @ T=0C in seawater
#ref_seagas = np.array([20, 349, 616]) / 985

def seawater_script(tempC, salinity):
    solutes = ref_solutes * (salinity / 35.2)
    working_script = f"""
    SOLUTION 1 Seawater-like composition
        temp      {tempC}      # Starting temperature in °C (to be changed each run)
        pH        8.0
        pressure  1
        water     1  # kg
        units     mol/kgw
        Na        {solutes[0]} #22.98977 g/mol
        Cl        {solutes[1]} #35.453
        Mg        {solutes[2]} #24.305
        SO4       {solutes[3]} #96.0626
        Ca        {solutes[4]} #40.078
        K         {solutes[5]} #39.0983
    """
    working_script += open('phreeqc-scripts/0813_seawater_2eq_phases.txt','r').read()
    #working_script += f"REACTION_TEMPERATURE 1\n    {loT}  {hiT}  in  {Tsteps} steps"
    working_script += open('phreeqc-scripts/0813_seawater_4output.txt','r').read()
    return working_script
```

```python
phreeqc.run_string(seawater_script(-3.5, 20))
selected_output = phreeqc.get_selected_output_array()
pd.DataFrame(selected_output[1:], columns=selected_output[0]).iloc[:,5:]
```

```python
#initial_conditions = open("frezchem-master/PHREEQC_Frz_v2.txt", 'r').read()
temps, salis = np.linspace(-30, 20, 100), np.linspace(0.1, 300.1, 300)
temps, salis = np.meshgrid(temps, salis, indexing='ij')

eq_salis, icefrac = np.zeros_like(temps), np.zeros_like(temps)
for i in range(icefrac.shape[0]):
    for j in range(icefrac.shape[1]):
        try:
            phreeqc.run_string(seawater_script(temps[i,j], salis[i,j]))
        except:
            print('failure')
            icefrac[i,j] = 1.0
            eq_salis[i,j] = -1.0
        else:
            output = phreeqc.get_selected_output_array()
            tmp1 = output[0].index('m_H2O(mol/kgw)')
            tmp2 = np.array([output[2][i] for i in range(len(output[0])) if ('(mol/kgw)' in output[0][i] and i != tmp1)])
            eq_salis[i,j] = np.sum(tmp2 * ref_molmasses) * 35.2/34.8625
            icefrac[i,j] = output[2][output[0].index('Ice(s)')] / output[2][tmp1]
#pd.DataFrame(selected_output[1:], columns=selected_output[0])
```

```python
#eq_salis * (1-icefrac)
#selected_output["Ice(s)"]/selected_output["m_H2O(mol/kgw)"]
plt.imshow(icefrac, aspect=8, extent=[salis[0,0], salis[0,-1], temps[-1,0], temps[0,0]])
plt.xlabel('absolute salinity (g/kg)')
plt.ylabel('T (°C)')
plt.colorbar(label='ice fraction')
```

```python
import pytzer as pz
pz.set_library(pz, "CWTD23")

solutes = pz.get_solutes() #https://pytzer.hseao3.group/model/ - don't forget DOI reference!
solutes['Cl'] = 0.546
solutes['Na'] = 0.469
solutes['Mg'] = 0.0528
solutes['SO4'] = 0.0282
solutes['Ca'] = 0.0103
solutes['K'] = 0.0102
solutes_sum = np.sum(np.array(list(solutes.values())))
solutes = {key:solutes[key]/solutes_sum for key in solutes.keys()}
#Gibbs_lo = pz.Gibbs_nRT(solutes, 17, 10)
#Gibbs_hi = pz.Gibbs_nRT(solutes, 290, 10)

temps = np.linspace(17, 290, 100)
salis = np.linspace(0, 5, 50)
#temps, salis = np.meshgrid(temps, salis, indexing='ij')
results = np.zeros((len(temps), len(salis)))
for i in range(len(temps)):
    for j in range(len(salis)):
        ssolutes = {key:solutes[key]*salis[j] for key in solutes.keys()}
        results[i,j] = pz.Gibbs_nRT(ssolutes, temps[i], 10)

plt.imshow(results); plt.show()
plt.plot(temps, results[:,10])
```
