```python
def bvspline_brineSpecificHeatCap(Tmin=240, Tmax=340, Tsteps=200, Smin=0, Smax=200, Ssteps=200):
    # brine heat capacity - https://web.mit.edu/lienhard/www/papers/journal/Thermophysical_properties_of_seawater-DWT-16-354-2010.pdf
    T = np.linspace(Tmin, Tmax, Tsteps)
    S = np.linspace(Smin, Smax, Ssteps)
    T, S = np.meshgrid(T, S, indexing='ij')
    
    A = 5.328 - 9.76e-2*S + 4.04e-4*S**2
    B = -6.913e-3 + 7.351e-4*S - 3.15e-6*S**2
    C = 9.6e-6 - 1.927e-6*S + 8.23e-9*S**2
    D = 2.5e-9 + 1.666e-9*S - 7.125e-12*S**2
    
    c = A + B*T + C*T**2 + D*T**3
    
    return scipy.interpolate.RectBivariateSpline(T[:,0], S[0,:], c).ev

def uvspline_iceSpecificHeatCap(Tmin=240, Tmax=340, Tbins=200):
    #ice heat capacity - [whoops forgot source]
    T = np.linspace(Tmin, Tmax, Tbins)
    c = T * 2106 / 273.15
    return scipy.interpolate.make_splrep(T, c)

def uvspline_brineFreezingEnthalpy():
    '''returns enthalpy of freezing as a function of salinity.'''
    #specific enthalpy measurements - https://www.sciencedirect.com/science/article/pii/S2666202722000398
    x = np.array([0.5, 1, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0, 4.5]) #salinity
    y = np.array([317.02, 313.83, 307.39, 297.14, 295.18, 292.32, 289.62, 287.54, 282.44])*1000 #J/g to J/kg
    return scipy.interpolate.make_splrep(x, y)

import PhreeqcSolver
#import importlib
#importlib.reload(PhreeqcSolver)
def bvsplines_brinePhaseOps(Tmin=240, Tmax=340, Tsteps=200, Smin=0, Smax=200, Ssteps=200, is_kelvin=True, k=5):
    T = np.linspace(Tmin, Tmax, Tsteps)
    if (is_kelvin): T -= 273.15
    S = np.linspace(Smin, Smax, Ssteps)
    T, S = np.meshgrid(T, S, indexing='ij')
    ifrac = np.zeros(T.shape)
    gasSat = np.zeros(T.shape)

    solver = PhreeqcSolver.frezchem_seawater()
    for i in range(ifrac.shape[0]):
        for j in range(ifrac.shape[1]):
            ifrac[i,j] = solver.solve_at(T[i,j], S[i,j])[0]
    if (is_kelvin): T += 273.15
    #print(ifrac, T, S)
    ifrac = scipy.interpolate.RectBivariateSpline(T[:,0], S[0,:], ifrac, s=1, kx=k, ky=k)
    gasSat = scipy.interpolate.RectBivariateSpline(T[:,0], S[0,:], gasSat, s=1, kx=k, ky=k) #placeHOlder
    return (ifrac.ev, gasSat.ev)

def bvspline_ibmixEntropyAndSensibleHeatAndHeatCapacity(Tmin=240, Tmax=340, Tsteps=200, Smin=0, Smax=200, Ssteps=200, k=5):
    T = np.linspace(Tmin, Tmax, Tsteps)
    S = np.linspace(Smin, Smax, Ssteps)
    T, S = np.meshgrid(T, S, indexing='ij')

    iceFrac = bvsplines_brinePhaseOps(Tmin, Tmax, Tsteps, Smin, Smax, Ssteps)[0](T.ravel(),S.ravel()).reshape(T.shape)
    np.clip(iceFrac, 0, 1)
    bmask = iceFrac < 1
    brinSal = np.zeros(T.shape)
    brinSal[bmask] = S[bmask] / (1 - iceFrac[bmask])

    ic = uvspline_iceSpecificHeatCap(Tmin, Tmax)(T).reshape(T.shape)
    bc = bvspline_brineSpecificHeatCap(Tmin, Tmax, Tsteps, Smin, Smax, Ssteps)(T, S).reshape(T.shape)
    ibfe = uvspline_brineFreezingEnthalpy()(S).reshape(T.shape)
    mix_cp = ic * iceFrac + bc * (1 - iceFrac)

    delT = (Tmax-Tmin)/Tsteps
    delQI = (iceFrac[1:,:] - iceFrac[:-1,:])*ibfe[:-1,:]
    delQ = mix_cp * delT
    delQ[1:,:] += delQI
    delS = delQ / T

    entropy, heat = np.cumsum(delS, axis=0), np.cumsum(delQ, axis=0)

    entropy = scipy.interpolate.RectBivariateSpline(T[:,0], S[0,:], entropy, kx=k, ky=k)
    heat = scipy.interpolate.RectBivariateSpline(T[:,0], S[0,:], heat, kx=k, ky=k)
    mix_cp = scipy.interpolate.RectBivariateSpline(T[:,0], S[0,:], mix_cp, kx=k, ky=k)
        
    return entropy.ev, heat.ev, mix_cp.ev

def uvspline_porosityPermeability(porosityMin=0, porosityMax=1, porositySteps=200):
    # percolation model in young sea ice - https://presentations.copernicus.org/EGU21/EGU21-3510_presentation.pdf
    # thermal evolution of permeability - https://agupubs.onlinelibrary.wiley.com/doi/full/10.1029/2007GL030447
    # equation 4 - hierarchical model approx : k = 3e-8 * (porosity - porosity_critical)**2 where critical porosity ~ 0.05
    # equation 5 - hierarchical model : k = 3e-8 * porosity**3
    # brine flow - Permeability of young sea ice from microtomographic images - https://core.ac.uk/download/pdf/30897595.pdf
    # https://www.cambridge.org/core/journals/annals-of-glaciology/article/chlorinitysalinity-distribution-patterns-in-experimental-granular-sea-ice/3F44FD5D464E5004DBBE896804354A6D
    # .. is identically dependent on depth in all experiments, independent of the initial thermal driving. If the Reference Wakatsuchi and SaitoWakatsuchi and Saito (1985)
    #    relationship holds, this means that equivalent porosities are achieved with different channel configurations, these configurations depending on the initial freezing
    #    conditions

    porosity = np.linspace(porosityMin, porosityMax, porositySteps)
    permeability = 3e-8 * porosity**3

    permeability = scipy.interpolate.make_splrep(porosity, permeability)
    return permeability
    
```

```python
# Moore neighborhood
i_nb = np.array([-1,0,1,-1,1,-1,0,1])
j_nb = np.array([-1,-1,-1,0,0,1,1,1])

def update(temps, brine, icmat, timeStep=1, surf_temp=258, sea_temp=271.35, is_first_cycle=False):
    if (not is_first_cycle):
        brine[0] = brine[1].copy()
        temps[0] = temps[1].copy()
        icmat[0] = icmat[1].copy()
    
    temps[0,:2,:] = surf_temp + 3*(np.random.rand(1, width) - 0.5)
    #temps[temps > sea_temp+5] = sea_temp+5
    delQ = np.zeros((depth,width)) # joules
    delW = np.zeros((depth,width)) # kg of water
    delS = np.zeros((depth,width)) # grams of salt
    delI = np.zeros((depth,width)) # kg of ice
    delTime = timeStep

    tmp = icmat[0,:,:,1] > 0
    porosity = icmat[0,:,:,0].copy()
    porosity[tmp] = 1 - porosity[tmp] / icmat[0,:,:,1][tmp] * 0.9167
    porosity[~tmp] = 1
    #waterDepth = int(depth * (1 - np.sum(icmat[0,:,:,0] + brine[0,:,:,0] + 1e-3*brine[0,:,:,1]) / (width*depth*1.030)))
    
    #touchesSea = brine[0,:,:,0] - icmat[0,:,:,0] > 0.99999
    #touchesSea[-1,:] = True
    #for i in reversed(range(1, depth)):
    #    touchesSea[i-1] = touchesSea[i] * touchesSea[i-1]
    #touchesSea[:4] = False
    #temps[0][touchesSea] = sea_temp
    #brine[0][touchesSea][1] = 35*brine[0][touchesSea][0]
    temps[0,-2:,:] = sea_temp + 0.15*(np.random.rand(1, width) - 0.5)
    brine[0,-2:,:,1] = 35#*brine[0,-10:,:,0]
    brine[0,-2:,:,0] = 1
    icmat[0,-2:,:,0] = 0
    icmat[0,-2:,:,1] = 1e-8

    permeability = poroPerm_ev(porosity.ravel()).reshape(porosity.shape)
    #Pg = (brine[0,i,j,1]*1e-3 + brine[0,i,j,0]) * 9.80665
    # Hydrostatic pore water pressure revisited - Stewart & Albertz, 2022 : https://onlinelibrary.wiley.com/doi/10.1111/bre.12708
    # i.e. i gotta do my own thing i guess!!! (pain)
    #head_hgt = porosity[i] > 0.05
    #for i in range(1, depth):  
    #P += icmat[0,:,:,0]

    h2omass = brine[0,:,:,0] + icmat[0,:,:,0]
    ifrac = bpo_ev[0](temps[0].ravel(), (brine[0,:,:,1]/h2omass).ravel()).reshape(h2omass.shape)
    np.clip(ifrac, 0, 1)
    delI = ((ifrac - icmat[0,:,:,0] / h2omass) * h2omass)
    entr, enth, cap = mixESHC[0](temps[0], brine[0,:,:,1]/h2omass), mixESHC[1](temps[0], brine[0,:,:,1]/h2omass), mixESHC[2](temps[0], brine[0,:,:,1]/h2omass)
    cap[cap < 4.184] = 4.184
    cap *= cell_dist
    enth[enth < 0] = 500
    entr[entr < 0] = 500 / temps[0][entr < 0]
    lattice_param = ifrac * 5.9 + (1 - ifrac) * 2.77 #lattice distances, water from https://www.frontiersin.org/journals/physics/articles/10.3389/fphy.2014.00022/full
    for i in range(1, depth-1):
        for j in range(1, width-1):
            #transfer operations
            i_nbl, j_nbl = i+i_nb, j+j_nb
            tmp = (j_nbl > 0) * (j_nbl < width)
            i_nbl, j_nbl = i_nbl[tmp], j_nbl[tmp] #don't interact with horizontal edges
            
            nb_bl = brine[0, i_nbl, j_nbl].reshape(i_nbl.size,3)
            nb_tl = temps[0, i_nbl, j_nbl].reshape(i_nbl.size)
            nb_il = icmat[0, i_nbl, j_nbl].reshape(i_nbl.size, 2)
            nb_pr = porosity[i_nbl, j_nbl].reshape(i_nbl.size)
            nb_cp = cap[i_nbl, j_nbl].reshape(i_nbl.size)
            nb_q = enth[i_nbl, j_nbl].reshape(i_nbl.size)
            networked = porosity[i,j] > 0.05
            for k in range(i_nbl.size):
                #heat conduction
                llattice_param = ( lattice_param[i,j] + lattice_param[i_nbl[k], j_nbl[k]] ) / 2
                a = 0.5*(1/temps[0,i,j] - 1/nb_tl[k])**2 / ((1/temps[0,i,j])**2/cap[i,j] + (1/nb_tl[k])**2/nb_cp[k]) #delS_eq
                a = 0.5 * (cell_dist * llattice_param)**2 * (1/temps[0,i,j]) * np.exp(-a) #diffusivity
                delQ[i,j] += -(a * (enth[i,j] - nb_q[k]) / (cell_dist)) * h2omass[i,j] #Fick's first law, H2Omass as cross-section
                
                #brine transfer
                if (networked and nb_pr[k] > 0.05):
                    salr = np.linspace(brine[0,i,j,1]/brine[0,i,j,0], nb_bl[k,1]/nb_bl[k,0], 20)
                    if (salr[0] != salr[1]):
                        t,a = np.meshgrid(np.array([temps[0,i,j],nb_tl[k]]), salr, indexing='ij')
                        entrs = mixESHC[0](t.ravel(),a.ravel()).reshape(t.shape)
                        entrs[0] *= brine[0,i,j,0]
                        entrs[1] *= nb_bl[k,0]
                        dS_dSal = (entrs[0,1] - entrs[0,0]) / (salr[1] - salr[0])
                        #print(np.sum(entrs, axis=0))
                        a = entrs[0, np.argmax(np.sum(entrs, axis=0))] - entrs[0,0] #delS_eq
                        #if (np.abs(a) > 0): print(a)
                        a = 0.5 * (cell_dist**2 * dS_dSal * np.exp(-a)) #diffusivity
                        a = -(a * (salr[0] - salr[-1]) / (cell_dist)) * porosity[i,j]# 0.5*(permeability[i_nbl[k],j_nbl[k]] + permeability[i,j])
                        #if (a > 0): print(a)
                        delS[i,j] += a
                #    dgrav = (nb_bl[k, 1] - brine[0,i,j,1])*1e-3 * i_nb[k]*0.001*cell_dist*9.81 #joules
                    #delQ[i,j] += (783.07*((nb_tl[k]+temps[0,i,j])/2)**(-1.044)) * (nb_bl[k,0]) * (nb_tl[k]-temps[0,i,j]) / cell_dist
            
            #if (temps[0,i,j] + delQ[i,j] / cap[i,j] < surf_temp - 50):
            #    delTime -= delTime * (surf_temp - temps[1,i,j])/(delQ[i,j]/cap[i,j]+1) / (depth*width)

    # STABILITY CONTROL
    delT = delQ/cap
    #tmp = np.min(delT / temps[0])
    #if (tmp < -0.1): delTime /= -(20*tmp)
    #tmp = np.max(delT / temps[0])
    #if (tmp > 0.1): delTime /= (20*tmp)
    tmp = np.max(np.abs(delI / h2omass))
    if (tmp > 0.1): delTime /= (20*tmp)
    tmp = np.min((delS / (brine[0,:,:,1]+1e-8)))
    if (tmp < -0.1): delTime /= -(20*tmp)

    # UPDATES
    temps[1] = temps[0] + delT * delTime
    brine[1,:,:,0] = brine[0,:,:,0] + delW * delTime
    brine[1,:,:,1] = brine[0,:,:,1] + delS * delTime
    #delI[delI<0] = -np.min(np.stack((-delI, icmat[0,:,:,0]/delTime)), axis=0)[delI<0]
    #delI[delI>0] = np.min(np.stack((delI, brine[0,:,:,0]/delTime)), axis=0)[delI>0]
    brine[1,:,:,0] = brine[0,:,:,0] - delI * delTime
    icmat[1,:,:,0] = icmat[0,:,:,0] + delI * delTime
    icmat[1,:,:,1] = icmat[0,:,:,1] + delI / 0.9167 * delTime
    #print (delTime)

    return temps, brine, icmat, delQ, cap
```

```python
#REFERENCES
# ice heat capacity - https://www.researchgate.net/figure/Specific-heat-capacity-at-constant-pressure-for-ice-Ih-solid-line-corresponding-to_fig17_259742860
# brine heat capacity - https://web.mit.edu/lienhard/www/papers/journal/Thermophysical_properties_of_seawater-DWT-16-354-2010.pdf
# brine phase evolution - calculated with PHREEQC
mixESHC = bvspline_ibmixEntropyAndSensibleHeatAndHeatCapacity(240, 280, 40, 0, 100, 25)
bpo_ev = bvsplines_brinePhaseOps(240, 280, 40, 0, 100, 25, is_kelvin=True)
poroPerm_ev = uvspline_porosityPermeability()
```

```python
depth = 50 # cm
width = 10  # cm
cell_dist = 0.01 #m

steps = 10
#grid = np.random.choice([0, 1], size=(depth, width))
#gases = np.zeros((2, depth, width)) # litres

wtemps = np.zeros((2, depth, width)) # Kelvin scale!
wtemps[0,:,:] = 260

wbrine = np.zeros((2, depth, width, 3))
wbrine[0,:,:,0] = 1 - 1e-8 #1 kg water
wbrine[0,:,:,1] = 10 #10 g salt
wbrine[0,:,:,2] = 0.001 #0.001 mol/kg dissolved gas

wicmat = np.zeros((2, depth, width, 2))
wicmat[0,:,:,0] = 1 - wbrine[0,:,:,0] #0 kg ice
wicmat[0,:,:,1] = 1#wicmat[0,:,:,0]/0.9167 #0 litres volume
#note - pure ice density : 0.9167 kg/L

ist = 250
ibt = 271.35

#salinity_sea_ice [ppt] 	depth [m] 	temperature_sea_ice [K]
new_input = False
if (new_input):
    tmp = np.array([[0.000e+00, 6.500e-02, 2.623e+02],
           [0.000e+00, 1.800e-01, 2.628e+02],
           [3.000e-01, 2.800e-01, 2.633e+02],
           [8.000e-01, 3.800e-01, 2.637e+02],
           [1.800e+00, 4.800e-01, 2.641e+02],
           [2.600e+00, 5.800e-01, 2.645e+02],
           [3.000e+00, 6.800e-01, 2.650e+02],
           [3.600e+00, 7.800e-01, 2.654e+02],
           [4.200e+00, 8.800e-01, 2.658e+02],
           [3.700e+00, 9.800e-01, 2.663e+02],
           [4.000e+00, 1.080e+00, 2.667e+02],
           [4.500e+00, 1.180e+00, 2.671e+02],
           [4.400e+00, 1.280e+00, 2.675e+02],
           [4.300e+00, 1.380e+00, 2.680e+02],
           [4.000e+00, 1.480e+00, 2.684e+02],
           [3.700e+00, 1.580e+00, 2.688e+02],
           [4.200e+00, 1.680e+00, 2.693e+02],
           [4.400e+00, 1.780e+00, 2.697e+02],
           [2.800e+00, 1.880e+00, 2.701e+02],
           [1.900e+00, 1.980e+00, 2.706e+02],
           [3.200e+00, 2.080e+00, 2.710e+02],
           [3.700e+00, 2.180e+00, 2.714e+02]])
    depths = (tmp[:,1]*100).astype(int) #m to cm
    temp_samp = scipy.interpolate.make_splrep(depths, tmp[:,2])
    sali_samp = scipy.interpolate.make_splrep(depths, tmp[:,0])
    ist = np.min(tmp[:,2])
    ibt = np.max(tmp[:,2])
    depth_max = np.max(tmp[:,1]*100)
    
    depths_grid = np.arange(depth_max).astype(int)
    temp_samp = temp_samp(depths_grid)
    sali_samp = sali_samp(depths_grid)
    ibf = bpo_ev[0](temp_samp, sali_samp)
    wicmat[0,depths_grid,:,1] = (ibf / 0.9167)[:, np.newaxis]
    wicmat[0,depths_grid,:,0] = (ibf)[:, np.newaxis]
    wbrine[0,depths_grid,:,0] = (1 - ibf)[:, np.newaxis]
    wbrine[0,depths_grid,:,1] = sali_samp[:, np.newaxis]
    wtemps[0,:,:] = ibt
    wtemps[0,depths_grid,:] = temp_samp[:, np.newaxis]

    plt.imshow(wicmat[0,:,:,0].T); plt.colorbar(shrink=0.15); plt.show()
    plt.imshow(wtemps[0,:,:].T); plt.colorbar(shrink=0.15); plt.show()
    plt.imshow(wbrine[0,:,:,1].T); plt.colorbar(shrink=0.15); plt.show()

first_run = True
```
